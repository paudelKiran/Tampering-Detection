```python
import tensorflow as tf
import numpy as np
import time
```

```python
print("TensorFlow version:", tf.__version__)
print("GPU devices:", tf.config.list_physical_devices('GPU'))
print("Num GPUs Available:", len(tf.config.list_physical_devices('GPU')))
```

```python
# Create large matrices for testing
matrix_size = 5000
iterations = 5

print(f"\nTest: Matrix multiplication ({matrix_size}x{matrix_size})")
print(f"Iterations: {iterations}\n")

# GPU Test
print("GPU Performance Test")
print("-" * 60)
with tf.device('/GPU:0'):
    # Warm up
    a_gpu = tf.random.normal([matrix_size, matrix_size])
    b_gpu = tf.random.normal([matrix_size, matrix_size])
    _ = tf.matmul(a_gpu, b_gpu)
    
    # Actual timing
    start_time = time.time()
    for i in range(iterations):
        result_gpu = tf.matmul(a_gpu, b_gpu)
    gpu_time = time.time() - start_time
    
print(f"GPU Time: {gpu_time:.4f} seconds")
print(f"Average per iteration: {gpu_time/iterations:.4f} seconds\n")

# CPU Test
print("CPU Performance Test")
print("-" * 60)
with tf.device('/CPU:0'):
    # Warm up
    a_cpu = tf.random.normal([matrix_size, matrix_size])
    b_cpu = tf.random.normal([matrix_size, matrix_size])
    _ = tf.matmul(a_cpu, b_cpu)
    
    # Actual timing
    start_time = time.time()
    for i in range(iterations):
        result_cpu = tf.matmul(a_cpu, b_cpu)
    cpu_time = time.time() - start_time
    
print(f"CPU Time: {cpu_time:.4f} seconds")
print(f"Average per iteration: {cpu_time/iterations:.4f} seconds\n")

# Comparison
print("="*60)
print("RESULTS")
print("="*60)
print(f"GPU Time: {gpu_time:.4f} seconds")
print(f"CPU Time: {cpu_time:.4f} seconds")
print(f"Speedup: {cpu_time/gpu_time:.2f}x faster on GPU")
print(f"Performance gain: {((cpu_time - gpu_time) / cpu_time * 100):.1f}%")
```
